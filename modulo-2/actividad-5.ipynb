```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Configuración estética
plt.style.use('ggplot')
sns.set(font_scale=1.1)

# Cargamos los datos
df = pd.read_csv('CAR DETAILS FROM CAR DEKHO.csv')
df.head()


# ¿Cuál es el promedio de autos listados por año? Graficá la cantidad de autos por año?
print('Cantidad media de autos listados por año: ', df['year'].value_counts().mean().__round__(2))
autos_x_anio = df['year'].value_counts().sort_index()
plt.figure(figsize=(8,5))
sns.barplot(x=autos_x_anio.index, y=autos_x_anio.values, hue=autos_x_anio.values, dodge=False, legend=False,  palette='viridis')
plt.title('Cantidad de autos listados por año')
plt.xlabel('Año')
plt.ylabel('Cantidad de autos')
plt.xticks(rotation=45)
plt.show()
```

```python
# Calculá la media, mediana y moda del `year` (año del vehículo).
print('Media del año del vehículo: ', int(df['year'].mean()))
print('Mediana del año del vehículo: ', int(df['year'].median()))
print('Moda del año del vehículo: ', df['year'].mode()[0])
```

```python
# Realizá un boxplot del kilometraje recorrido (`km_driven`) para detectar outliers.
plt.figure(figsize=(8,5))
sns.boxplot(x=df['km_driven'])
plt.title('Boxplot del Kilometraje Recorrido')
plt.show()
```

```python
# Calculá el rango, varianza y desviación estándar del `year`
year_range = df['year'].max() - df['year'].min()
year_variance = df['year'].var()    
year_std = df['year'].std()

print(f"Rango: {year_range}")
print(f"Varianza: {year_variance.round(2)}")
print(f"Desviación Estándar: {year_std.round(2)}")
```

```python
# Hacé un histograma del `selling_price` para autos que usan combustible "Diesel".
diesel_cars= df[df['fuel'] == 'Diesel']
plt.figure(figsize=(8,5))
sns.histplot(diesel_cars['selling_price'], bins=20, kde=True, color='blue')
plt.title('Histograma del Precio de Venta para Autos Diesel')
plt.xlabel('Precio de Venta')
plt.ylabel('Frecuencia')
plt.show()
```

```python
# Graficá la distribución de autos por tipo de transmisión (`manual` vs `automatic`).
transmission_counts = df['transmission'].value_counts() 
plt.figure(figsize=(6,6))
plt.pie(transmission_counts, labels=transmission_counts.index, autopct='%1.1f%%')
plt.title('Distribución de autos por tipo de transmisión')
plt.show()
```

```python
# ¿Cuál es la relación entre el tipo de combustible (`fuel`) y el precio promedio de venta?
relacion = df.groupby('fuel')['selling_price'].mean().reset_index()
plt.figure(figsize=(8,5))
sns.barplot(x='fuel', y='selling_price', data=relacion, hue='fuel', palette='muted')
plt.title('Relación entre Tipo de Combustible y Precio Promedio de Venta')
plt.xlabel('Tipo de Combustible')   
plt.ylabel('Precio Promedio de Venta')
plt.show()
```

```python
# Usando un `groupby`, obtené el precio medio por año de fabricación.
precio_medio_x_anio = df.groupby('year')['selling_price'].mean().reset_index()
plt.figure(figsize=(10,6))
sns.lineplot(x='year', y='selling_price', data=precio_medio_x_anio, marker='o')
plt.title('Precio Medio de Venta por Año de Fabricación')
plt.xlabel('Año de Fabricación')
plt.ylabel('Precio Medio de Venta')
plt.grid()
plt.show()
```

```python
# Mostrá en un gráfico de barras cuántos autos pertenecen a cada número de dueños (`owner`).
owner_counts = df['owner'].value_counts()
plt.figure(figsize=(8,5))
sns.barplot(x=owner_counts.index, y=owner_counts.values, hue=owner_counts.values, palette='viridis')
plt.title('Cantidad de Autos por Número de Dueños')
plt.xlabel('Número de Dueños')
plt.ylabel('Cantidad de Autos')
plt.xticks(rotation=45, fontsize=9)
plt.show()
```

```python
# Calculá las medidas de dispersión para el precio de autos automáticos únicamente.
automatico = df[df['transmission'] == 'Automatic']
rango_auto = automatico['selling_price'].max() - automatico['selling_price'].min()
varianza_auto = automatico['selling_price'].var()
desviacion_auto = automatico['selling_price'].std()
print(f"Rango del precio de autos automáticos: {rango_auto}")
print(f"Varianza del precio de autos automáticos: {varianza_auto.round(2)}")
print(f"Desviación estándar del precio de autos automáticos: {desviacion_auto.round(2)}")
```
